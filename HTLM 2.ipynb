```python
import pandas as pd
import numpy as np

data = {
    'department': np.random.choice(['Operations', 'Finance', 'HR', 'Logistics'], 50),
    'conflict_cause': np.random.choice(['Communication', 'Resource Allocation', 'Leadership'], 50),
    'score': np.random.randint(1, 10, 50)
}
df = pd.DataFrame(data)
df.to_csv('pad_conflicts.csv', index=False)

df = pd.read_csv('pad_conflicts.csv')
df.head()
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nMost common conflict cause:")
print(df['conflict_cause'].value_counts())

print("\nAverage score by department:")
print(df.groupby('department')['score'].mean())
```

```python
import matplotlib.pyplot as plt

df['conflict_cause'].value_counts().plot(kind='bar')
plt.title('PAD Conflicts by Cause - Brent Analysis')
plt.ylabel('Count')
plt.show()

df.groupby('department')['score'].mean().plot(kind='bar', color='orange')
plt.title('Avg Conflict Score by Department')
plt.show()
```
